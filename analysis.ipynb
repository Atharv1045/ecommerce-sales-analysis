```python
# E-commerce Sales Analysis

## Week 4 – Data Visualization & Complete Data Analysis Project

This notebook performs a complete analysis of an e-commerce sales dataset using Python, Pandas, and Matplotlib.

### Objectives

- Load sales data from a CSV file.
- Explore the dataset.
- Check and clean the data.
- Calculate important sales metrics.
- Analyze sales by product.
- Analyze sales by region.
- Analyze monthly sales trends.
- Create bar and line charts.
- Extract meaningful business insights.
```

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
## 1. Load the Dataset

The sales dataset is stored in the `data` folder. Pandas is used to read the CSV file into a DataFrame.
```

```python
df = pd.read_csv("data/sales_data.csv")

print("Dataset loaded successfully.")
print(f"Rows: {df.shape[0]}")
print(f"Columns: {df.shape[1]}")

df.head()
```

```python
## 2. Explore the Dataset

The dataset is explored to understand its size, columns, data types, and initial records.
```

```python
print("Dataset Shape:")
print(df.shape)

print("\nColumn Names:")
print(df.columns.tolist())

print("\nData Types:")
print(df.dtypes)

print("\nDataset Information:")
df.info()
```

```python
print("First 5 Rows:")
display(df.head())

print("\nLast 5 Rows:")
display(df.tail())
```

```python
## 3. Data Cleaning

Before performing the analysis, the dataset is checked for missing values and duplicate records.

The `Date` column is converted to datetime format so that monthly sales trends can be calculated.
```

```python
print("Missing values before cleaning:")
display(df.isnull().sum())

print("Duplicate rows before cleaning:", df.duplicated().sum())
```

```python
# Convert Date column to datetime format
df["Date"] = pd.to_datetime(df["Date"], errors="coerce")

# Fill missing numerical values with the median
numeric_columns = df.select_dtypes(include="number").columns

for column in numeric_columns:
    if df[column].isnull().any():
        df[column] = df[column].fillna(df[column].median())

# Fill missing text values with the most common value
text_columns = df.select_dtypes(include="object").columns

for column in text_columns:
    if df[column].isnull().any():
        df[column] = df[column].fillna(df[column].mode()[0])

# Remove duplicate rows
df = df.drop_duplicates()

# Remove rows with invalid dates
df = df.dropna(subset=["Date"])

print("Data cleaning completed.")
print("Rows after cleaning:", len(df))
```

```python
print("Missing values after cleaning:")
display(df.isnull().sum())

print("Duplicate rows after cleaning:", df.duplicated().sum())
```

```python
## 4. Basic Sales Analysis

The following metrics are calculated:

- Total sales
- Average sale
- Highest sale
- Lowest sale
- Total quantity sold
```

```python
total_sales = df["Total_Sales"].sum()
average_sale = df["Total_Sales"].mean()
highest_sale = df["Total_Sales"].max()
lowest_sale = df["Total_Sales"].min()
total_quantity = df["Quantity"].sum()

print(f"Total Sales: {total_sales:,.2f}")
print(f"Average Sale: {average_sale:,.2f}")
print(f"Highest Sale: {highest_sale:,.2f}")
print(f"Lowest Sale: {lowest_sale:,.2f}")
print(f"Total Quantity Sold: {total_quantity}")
```

```python
## 5. Sales Analysis by Product

The sales data is grouped by product to determine which products generate the highest total sales.
```

```python
product_sales = (
    df.groupby("Product")["Total_Sales"]
    .sum()
    .sort_values(ascending=False)
)

display(product_sales.to_frame(name="Total Sales"))

best_product = product_sales.index[0]

print(f"Best-selling product: {best_product}")
print(f"Total sales generated: {product_sales.iloc[0]:,.2f}")
```

```python
### Bar Chart – Total Sales by Product

A bar chart is used to compare the total sales generated by each product.
```

```python
plt.figure(figsize=(9, 6))

plt.bar(product_sales.index, product_sales.values)

plt.title("Total Sales by Product")
plt.xlabel("Product")
plt.ylabel("Total Sales")
plt.xticks(rotation=0)

plt.tight_layout()
plt.show()
```

```python
### Product Insight

The bar chart makes it easy to compare product performance. Laptop is the best-performing product based on total sales.
```

```python
## 6. Sales Analysis by Region

Sales are grouped by region to identify the strongest and weakest performing regions.
```

```python
region_sales = (
    df.groupby("Region")["Total_Sales"]
    .sum()
    .sort_values(ascending=False)
)

display(region_sales.to_frame(name="Total Sales"))

best_region = region_sales.index[0]

print(f"Best-performing region: {best_region}")
print(f"Total sales generated: {region_sales.iloc[0]:,.2f}")
```

```python
### Bar Chart – Total Sales by Region
```

```python
plt.figure(figsize=(9, 6))

plt.bar(region_sales.index, region_sales.values)

plt.title("Total Sales by Region")
plt.xlabel("Region")
plt.ylabel("Total Sales")
plt.xticks(rotation=0)

plt.tight_layout()
plt.show()
```

```python
### Regional Insight

The regional analysis shows that North is the highest-performing region based on total sales, while West has the lowest total sales.
```

```python
## 7. Monthly Sales Trend

The Date column is used to group transactions by month. A line chart is then created to visualize how sales change over time.
```

```python
monthly_sales = (
    df.groupby(df["Date"].dt.to_period("M"))["Total_Sales"]
    .sum()
)

monthly_sales.index = monthly_sales.index.astype(str)

display(monthly_sales.to_frame(name="Total Sales"))
```

```python
### Line Chart – Monthly Sales Trend
```

```python
plt.figure(figsize=(9, 6))

plt.plot(
    monthly_sales.index,
    monthly_sales.values,
    marker="o"
)

plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)
plt.grid(True)

plt.tight_layout()
plt.show()
```

```python
### Monthly Sales Insight

The monthly sales chart shows how revenue changes over time. March 2024 recorded the highest monthly sales, while April 2024 recorded the lowest monthly sales.
```
